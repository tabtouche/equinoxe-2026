# Données publiques SCHL — Montréal et Ottawa, 2021–2025

Ce notebook construit `df_public` à partir des cinq fichiers Excel de `raw_public_data`. Une ligne correspond à un marché et à une année d'enquête (octobre). Montréal représente le contexte régional des immeubles québécois; Ottawa correspond à la **partie ontarienne** de la RMR Ottawa-Gatineau, et non à Gatineau.

| Variable | Définition et unité | Tableau source |
|---|---|---|
| `vacancy_rate` | Taux d'inoccupation, en %; 2,1 signifie 2,1 % | 1.0, année du fichier |
| `turnover_rate` | Taux de rotation/roulement des locataires, en % | 1.0, année du fichier |
| `rent` | Loyer mensuel moyen en dollars des appartements de **deux chambres**, immeubles neufs et existants | 1.0, année du fichier |
| `rental_universe` | Nombre total d'appartements visés par l'ELL dans le marché, toutes tailles | 4.1, univers des appartements visés par l'ELL |

Le périmètre est celui des immeubles locatifs d'initiative privée comptant au moins trois appartements. Nous ne mélangeons pas ces logements avec les appartements en copropriété offerts en location.

Nous retenons les valeurs de l'année courante de chaque fichier : 2022 provient du fichier 2022, même si un fichier ultérieur révise une valeur de 2022. Les millésimes restent ainsi explicites. Pour un backtest, vérifier aussi la date de publication effective des fichiers avant de les utiliser.
